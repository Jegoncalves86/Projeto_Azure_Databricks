```python
import os
import sys

# 1. Sobe um nível para encontrar a raiz do repositório Git no Databricks
raiz_repositorio = os.path.abspath("../..")

# 2. Adiciona a raiz do repositório no caminho de busca do Python
if raiz_repositorio not in sys.path:
    sys.path.append(raiz_repositorio)

# 3. Agora o import vai funcionar perfeitamente
from funcoes_diversas.funcoes_util import buscar_caminho
```

```python
resource_path = buscar_caminho("resource")
df_categories = spark.read.csv(f'{resource_path}/categories.csv', 
            header=True, 
            inferSchema=True,
            sep=',')
display(df_categories)
```

```python
# salvar o dataframe como camada brone
bronze_path = buscar_caminho("bronze")
df_categories.write\
    .mode('overwrite')\
    .format('delta')\
    .option('overwriteSchema', 'true')\
    .save(f'{bronze_path}/categories')
```
