```python
import os
import sys

# Procura automaticamente em qual pasta acima está a 'funcoes_diversas'
caminho = os.getcwd()
while caminho != os.path.dirname(caminho):
    if os.path.exists(os.path.join(caminho, "funcoes_diversas")):
        if caminho not in sys.path:
            sys.path.insert(0, caminho)
        break
    caminho = os.path.dirname(caminho)

from funcoes_diversas.funcoes_util import buscar_caminho
```

```python
resource_path = buscar_caminho("resource")
df_customers = spark.read.csv(f'{resource_path}/customers.csv', 
            header=True, 
            inferSchema=True,
            sep=',')
display(df_customers)
```

```python
# salvar o dataframe como camada brone
bronze_path = buscar_caminho("bronze")
df_customers.write\
    .mode('overwrite')\
    .format('delta')\
    .option('overwriteSchema', 'true')\
    .save(f'{bronze_path}/customers')
```
